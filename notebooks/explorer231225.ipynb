```python
stats_by_country = (
    merged_filtered_mapped
    .groupby(['countryname', 'bila_language_name_mapping'])[emotions]
    .agg(['mean', 'std'])
)
stats_by_country = pd.DataFrame(stats_by_country)
stats_by_country
```

```python
stats_by_country.columns = [
    f"{emotion}_{stat}" for emotion, stat in stats_by_country.columns
]
stats_by_country = stats_by_country.reset_index()
stats_by_country
```

```python
stats_long = (
    stats_by_country
    .set_index(['countryname', 'bila_language_name_mapping'])
    .filter(regex='_(mean|std)$')
    .stack()
    .reset_index()
)

stats_long[['word', 'stat']] = stats_long['level_2'].str.rsplit('_', n=1, expand=True)
stats_long = stats_long.rename(columns={0: 'value'}).drop(columns='level_2')

stats_long = (
    stats_long
    .pivot_table(
        index=['countryname', 'bila_language_name_mapping', 'word'],
        columns='stat',
        values='value'
    )
    .reset_index()
)

merged = elab_per_emotion.merge(
    stats_long,
    left_on=['langname', 'word'],
    right_on=['bila_language_name_mapping', 'word'],
    how='left'
)

merged
```

```python
merged = merged.merge(
    dictionary_means,
    left_on='id',
    right_on='id',
    how='left'
)
merged
```

```python
merged = merged.rename(columns={'mean': "response_mean"})
merged = merged.rename(columns={'std': "response_std"})
merged = merged.drop('bila_language_name_mapping', axis=1)
merged = merged[['langname', 'countryname', 'word', 'count', 'dictionary_count_mean', 'response_mean', 'response_std', 'id', 'year', 'glottocode' ]]
merged = merged.sort_values(by=["langname", 'word'])
merged
```

```python

import plotly.express as px

# 1. Prepare data for the Heatmap
# We'll pivot the data to show Languages on one axis and Emotion Words on the other
heatmap_data = merged.pivot_table(
    index='langname', 
    columns='word', 
    values='response_mean', 
    aggfunc='mean'
)

# Create the Interactive Heatmap
fig_heatmap = px.imshow(
    heatmap_data,
    labels=dict(x="Emotion Word", y="Language", color="Intensity (Mean)"),
    x=heatmap_data.columns,
    y=heatmap_data.index,
    color_continuous_scale='Viridis',
    title='<b>Cross-Linguistic Emotional Intensity Heatmap</b><br><sup>Average survey response mean for each emotion word by language</sup>',
    aspect="auto"
)

fig_heatmap.update_layout(
    title_font_size=20,
    xaxis_tickangle=-45,
    width=1000,
    height=800
)

# 2. Interactive Scatter Plot
# This helps visualize the relationship between word frequency and response intensity
fig_scatter = px.scatter(
    merged, 
    x='count', 
    y='response_mean', 
    color='word',
    size='dictionary_count_mean', # Size points by dictionary frequency
    hover_data=['langname', 'countryname'],
    title='<b>Word Frequency vs. Emotional Intensity</b>',
    labels={'count': 'Word Count', 'response_mean': 'Mean Response'},
    trendline="ols", # Adds a trend line to see general correlation
    template='plotly_white'
)

# To view these in a notebook or browser:
fig_heatmap.show()
fig_scatter.show()

# To save as interactive HTML files:
# fig_heatmap.write_html("emotion_heatmap.html")
# fig_scatter.write_html("emotion_scatter.html")
```

```python
import plotly.express as px


# 1. Calculate the 'Relative Usage Index'
merged['rel_usage_index'] = (merged['count'] / merged['dictionary_count_mean']) * 100

# 2. Create the Bubble Chart
fig = px.scatter(
    merged, 
    x='rel_usage_index', 
    y='response_mean',
    size='count',          # Larger bubbles = more survey responses
    color='response_std',  # Color shows agreement/disagreement
    hover_name='word',     # Show the emotion word on hover
    hover_data=['langname', 'countryname', 'count'],
    color_continuous_scale='RdBu_r', # Red for high variance, Blue for consensus
    title='<b>Emotional Salience: Usage Prominence vs. Intensity</b>',
    labels={
        'rel_usage_index': 'Relative Usage Index (Survey vs. Dict %)',
        'response_mean': 'Average Intensity (Mean)',
        'response_std': 'Response Variation (Std Dev)'
    },
    template='plotly_white'
)

# Optional: Use a log scale for X if some words have massive relative frequency
# fig.update_xaxes(type="log") 

fig.show()
```

```python

# 1. Calculate the 'Relative Usage Index'
merged['rel_usage_index'] = (merged['count'] / merged['dictionary_count_mean']) * 100
merged['log_count'] = log(merged['count'])

# 2. Create the Square, High-Contrast Bubble Chart
fig = px.scatter(
    merged, 
    x='rel_usage_index', 
    y='response_mean',
    size='count',          
    color='response_std',  
    hover_name='word',     
    hover_data=['langname', 'countryname'],
    # 'Viridis' is much easier to see than 'RdBu' on a white background
    color_continuous_scale='viridis', 
    opacity=0.85,          # Increased opacity so bubbles aren't faint
    title='<b>Emotional Salience: Usage Prominence vs. Intensity</b>',
    labels={
        'rel_usage_index': 'Relative Usage Index (%)',
        'response_mean': 'Average Intensity (Mean)',
        'response_std': 'Response Variation (Std Dev)'
    },
    template='plotly_white',
    # Setting equal width and height makes the plot square
    width=800,
    height=800
)

# 3. Add a border to markers to make them distinct
fig.update_traces(
    marker=dict(
        line=dict(width=1, color='black')
    ),
    selector=dict(mode='markers')
)

# Adjust layout for better readability
fig.update_layout(
    title_x=0.5,
    margin=dict(l=50, r=50, t=80, b=50),
    coloraxis_colorbar=dict(title="Variation (Std)")
)

fig.show()
```

```python
import plotly.express as px
import pandas as pd
import numpy as np

# Assuming your dataframe is named 'merged'
# 1. Calculate the 'Relative Usage Index'
merged['ellaboration_relative_to_mean'] = (merged['count'] / merged['dictionary_count_mean'])

# 2. Calculate Log of Count to normalize bubble sizes
# np.log1p is used to handle low counts gracefully (calculates ln(1+x))
merged['log_count'] = np.log1p(merged['count'])

# 3. Create the Square, High-Contrast Bubble Chart
fig = px.scatter(
    merged, 
    x='ellaboration_relative_to_mean', 
    y='response_std',
    size='log_count',       # Using log_count for more uniform bubble sizes
    color='response_mean',  
    hover_name='word',     
    hover_data=['langname', 'countryname', 'count'], # Keep raw 'count' in hover
    color_continuous_scale='viridis', 
    opacity=0.85,
    title='<b>Ellaboration_relative_to_mean (Log Scaled) vs. Response STD </b>',
    labels={
        'ellaboration_relative_to_mean': 'ellaboration_relative_to_mean (count/dictionary_count_mean)',
        'response_mean': 'Mean of responses',
        'response_std': 'STD of responses',
        'log_count': 'Log of Count'
    },
    template='plotly_white',
    width=800,
    height=800
)

# 4. Refine marker appearance
fig.update_traces(
    marker=dict(
        line=dict(width=1, color='black'),
        sizemode='area', # Ensures bubble size is proportional to the area
        sizeref=0.1      # Adjust this number if bubbles are still too large/small
    ),
    selector=dict(mode='markers')
)

# 5. Center title and adjust colorbar
fig.update_layout(
    title_x=0.5,
    margin=dict(l=50, r=50, t=80, b=50),
    coloraxis_colorbar=dict(title="Response_mean")
)

fig.show()
```
