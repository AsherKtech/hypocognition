```python
covid_to_bila_full_lang_name_mapping = pd.read_csv("/data/home/asher.katz/hypocognition/reference/covid_to_full_bila_lang_name_mapping.csv")
covid_to_bila_full_lang_name_mapping = covid_to_bila_full_lang_name_mapping.sort_values(by = "easy").reset_index(drop=True)
covid_to_bila_full_lang_name_mapping
```

```python
# we need to make some connections with the different chinese lables
#Beijing Mandarin, Yue Chinese, Hakka Chinese, Min Nan Chinese, Mandarin Chinese, Min Dong Chinese,
chinese = bila_full_emotions[bila_full_emotions['langname'].isin(['Beijing Mandarin', 'Yue Chinese', 'Hakka Chinese', 'Min Nan Chinese', 'Mandarin Chinese', 'Min Dong Chinese'])]
```

```python
chinese_table = (
    chinese
    .pivot_table(
        index="word",
        columns="langname",
        values="count",
        aggfunc="sum",
        fill_value=0
    )
    .reset_index()
    .rename(columns={"word": "Emotion"})
)
chinese_table
```

```python
chinese_table = pd.concat(
    [chinese_table, pd.DataFrame(np.nan, index=range(2), columns=chinese_table.columns)],
    ignore_index=True
)
chinese_table['Emotion'] = emotions
chinese_table
```

```python
chinese_table.to_csv('/data/home/asher.katz/data/hypocognition/output_results/chinese_table.csv')
```
